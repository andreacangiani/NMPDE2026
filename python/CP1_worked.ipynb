```python
# Compute the integral of the function x**2 using trapezoidal integration on interval [0,2] (using for loop)


import numpy as np
def f(x):
    return x**2

N=100
x=np.linspace(0,6,N+1)

inegral=0
for i in range(len(x)-1):
    h=x[i+1]-x[i]
    f_left=f(x[i])
    f_right=f(x[i+1])
    
    inegral+=0.5*h*(f_left+f_right)
    
print(inegral)
    

import sympy as sym

x = sym.symbols('x')

## Define the function
f = x**2

## Compute the exact integral
I_exact = sym.integrate(f, (x, 0, 6))

print(I_exact)
```

$\delta_{h} f (x)= \frac{f(x+h/2)-f(x-h/2)}{h}  \quad$    (CD)

Let us now fix a grid and compute the FD in matrix form

```python
a=0
b=1
n=10
h=1/(n)
x=np.linspace(a,b,n+1)
```

Implement FD in matrix form

```python
fd = np.zeros((n+1,n+1))

for i in range(n):
    fd[i,i] = -1
    fd[i,i+1] = 1

fd = fd/h
```

Let's check convergence!

```python
def fdM1(a,b,n):
    h=(b-a)/n
    fdM1=np.zeros((n+1,n+1))
    for i in range(n):
        fdM1[i,i]=-1
        fdM1[i,i+1]=1
    return fdM1/(h)
        
```

```python
t = sym.var('t')
fsym=sym.sin(t)
fdsym=fsym.diff(t,1)
f=sym.lambdify(t,fsym)
fd=sym.lambdify(t,fdsym)
npexp=8
a=0
b=np.pi/2
error=np.zeros(npexp)
nn=np.zeros(npexp)
for i in range(npexp):
    n=2**(i+1)
    h=(b-a)/n
    nn[i]=n
    x=np.linspace(a,b,n+1)
    A=fdM1(a,b,n)
    u=f(x)
    uhd=A.dot(u)
    error[i]=max(abs(uhd[1:n]-fd(x[1:n])))
    
print(error)

```

Print error

```python
print(error)
```

Plot error

```python
import matplotlib.pyplot as plt

plt.loglog(1/nn, error, color='blue', label='Error')
plt.loglog(1/nn, 1/nn, color='red', label='O(n⁻¹)')

plt.legend()
plt.show()
```
