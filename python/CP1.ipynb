Let's check convergence!

```python
def fd(a,b,n):
    h = (b-a) / n
    fd = np.zeros((n+1,n+1))

    for i in range(n):
        fd[i,i] = -1
        fd[i,i+1] = 1

    return fd / h
```

```python
import sympy as sym
import numpy as np
a=0
b=np.pi/2
t = sym.var('t')
my_f = sym.sin(t)
fsym = sym.lambdify(t, my_f)
fsym_x = sym.lambdify(t, my_f.diff(t,1))

no_expe = 2
error = np.zeros(no_expe)
nn = np.zeros(no_expe)

for i in range(no_expe):
    n = 2**(i+1)
    nn[i] = n
    x = np.linspace(a,b,n+1)
    fd_mat = fd(a,b,n)
    u_x = fd_mat.dot(fsym(x))[1:n]
    error[i] = max (abs(u_x-fsym_x(x)[1:n]))


print(fd_mat)
```
